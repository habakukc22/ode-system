```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import least_squares

# Dados experimentais
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([1.2, 2.1, 2.9, 4.2, 5.1, 5.8])

# Função que retorna os resíduos
def residuals(params, x, y):
    a, b, c = params
    return a*x**2 + x*b + c - y

# Chute inicial
p0 = [1.0, 1, 1]

# Ajuste
result = least_squares(residuals, p0, args=(x, y))

# Parâmetros encontrados
a, b, c = result.x

print(f"a = {a:.4f}")
print(f"b = {b:.4f}")
print(f"c = {c:.4f}")

# Curva ajustada
xfit = np.linspace(0, 5, 100)
yfit = a*xfit**2 + b*xfit+c

plt.scatter(x, y, label="Dados")
plt.plot(xfit, yfit, 'r', label="Ajuste")
plt.legend()
plt.show()
```
